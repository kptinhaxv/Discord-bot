````python
import discord
from discord.ext import commands
import asyncio
from google.colab import auth
import gspread
from google.auth import default
import re, getpass, nest_asyncio
from discord.ui import Button, View

nest_asyncio.apply()

ID_DA_PLANILHA = '1IfwyJNL_n8Z5QgVtfbZKbFfPbwUkKsfUnBrMmmgop7o'
NOME_DA_ABA = 'COMP 1'

print("🔄 INICIANDO AUTENTICAÇÃO DO GOOGLE DRIVE...")
auth.authenticate_user()
creds, _ = default()
gc = gspread.authorize(creds)
aba = gc.open_by_key(ID_DA_PLANILHA).worksheet(NOME_DA_ABA)
print("📊 Planilha carregada com sucesso!")

print("\n🔐 CONFIGURAÇÃO DO DISCORD:")
TOKEN = getpass.getpass('Cole o seu TOKEN aqui e dê ENTER (fica invisível): ')

intents = discord.Intents.default()
intents.members = True
intents.message_content = True
intents.guilds = True
bot = commands.Bot(command_prefix='!', intents=intents)

painel_ativo = None
mensagem_painel = None

class PainelConvocacao(View):
    def __init__(self, nome_conteudo, data_saida):
        super().__init__(timeout=None)
        self.nome_conteudo = nome_conteudo
        self.data_saida = data_saida
        self.confirmados = []
        self.encerrado = False

    def gerar_embed_atualizado(self):
        if self.encerrado:
            titulo = "🔒 CONVOCAÇÃO ENCERRADA"
            desc = "**Inscrições finalizadas!** O grupo foi chamado para a atividade.\n──────────────────────────────────"
            cor = 0x95a5a6 # Cor cinza de encerrado
        else:
            titulo = "⚔️ CONVOCAÇÃO PARA ATIVIDADE OFICIAL ⚔️"
            desc = "**Atenção combatentes!** Um novo conteúdo foi agendado.\nClique nos botões abaixo para gerenciar sua presença.\n──────────────────────────────────"
            cor = 0x2ecc71 # Cor verde ativa

        embed = discord.Embed(title=titulo, description=desc, color=cor)
        embed.add_field(name="📦 Conteúdo Marcado", value=f"`{self.nome_conteudo}`", inline=True)
        embed.add_field(name="📅 Horário de Saída", value=f"`{self.data_saida}`", inline=True)

        vagas = 60 - len(self.confirmados)
        embed.add_field(name="📊 Status do Grupo", value=f"✅ **Confirmados:** `{len(self.confirmados)}` | 🚪 **Vagas Livres:** `{vagas if not self.encerrado else 0}`", inline=False)

        texto = "```md\n" + "\n".join([f"{i:02d}. # {u.display_name}" for i, u in enumerate(self.confirmados, 1)]) + "\n```" if self.confirmados else "*Nenhum guerreiro confirmado.*"
        embed.add_field(name="👥 Lista de Confirmados na Escala", value=texto, inline=False)

        if not self.encerrado:
            embed.set_footer(text="Use !callout para alertar os confirmados e trancar a lista")
        else:
            embed.set_footer(text="Painel sincronizado e trancado • Google Sheets")
        return embed

    @discord.ui.button(label="Confirmar Presença", style=discord.ButtonStyle.green, custom_id="btn_confirmar", emoji="🛡️")
    async def confirmar(self, interaction: discord.Interaction, button: discord.Button):
        if self.encerrado: return
        usuario = interaction.user
        id_usuario = str(usuario.id)
        if any(u.id == usuario.id for u in self.confirmados):
            await interaction.response.send_message("ℹ️ Você já está na lista!", ephemeral=True)
            return
        await interaction.response.send_message("🔄 Salvando...", ephemeral=True)
        try:
            map_cols = [('Q', 'R', 17, 18), ('T', 'U', 20, 21), ('W', 'X', 23, 24)]
            salvou = False
            for col_id, _, _, _ in map_cols:
                for linha in aba.get(f'{col_id}6:{col_id}25'):
                    if linha and "".join(re.findall(r'\d+', str(linha).strip())) == id_usuario:
                        salvou = True; break
                if salvou: break
            if not salvou:
                for col_id, col_nome, id_num, nome_num in map_cols:
                    valores = aba.get(f'{col_id}6:{col_id}25')
                    while len(valores) < 20: valores.append([])
                    for idx in range(20):
                        if not valores[idx] or "".join(re.findall(r'\d+', str(valores[idx]).strip())) == "":
                            aba.update_cell(6 + idx, id_num, id_usuario)
                            aba.update_cell(6 + idx, nome_num, usuario.display_name)
                            salvou = True; break
                    if salvou: break
            if not salvou:
                await interaction.followup.send("❌ Vagas lotadas na planilha!", ephemeral=True)
                return
            self.confirmados.append(usuario)
            await interaction.message.edit(embed=self.gerar_embed_atualizado())
            msg = await interaction.channel.send(f"⚔️ **{usuario.display_name}** entrou na lista!")
            await asyncio.sleep(5); await msg.delete()
        except Exception as e:
            await interaction.followup.send(f"❌ Erro: {e}", ephemeral=True)

    @discord.ui.button(label="Remover Presença", style=discord.ButtonStyle.red, custom_id="btn_remover", emoji="❌")
    async def remover(self, interaction: discord.Interaction, button: discord.Button):
        if self.encerrado: return
        usuario = interaction.user
        id_usuario = str(usuario.id)
        if not any(u.id == usuario.id for u in self.confirmados):
            await interaction.response.send_message("⚠️ Você não está na lista.", ephemeral=True)
            return
        await interaction.response.send_message("🔄 Removendo...", ephemeral=True)
        try:
            map_cols = [('Q', 'R', 17, 18), ('T', 'U', 20, 21), ('W', 'X', 23, 24)]
            limpou = False
            for col_id, col_nome, id_num, nome_num in map_cols:
                for idx, sublista in enumerate(aba.get(f'{col_id}6:{"Q" if col_id == "Q" else ("T" if col_id == "T" else "W")}25')):
                    if sublista and "".join(re.findall(r'\d+', str(sublista).strip())) == id_usuario:
                        aba.update_cell(6 + idx, id_num, "")
                        aba.update_cell(6 + idx, nome_num, "")
                        limpou = True; break
                if limpou: break
            for u in self.confirmados:
                if u.id == usuario.id: self.confirmados.remove(u); break
            await interaction.message.edit(embed=self.gerar_embed_atualizado())
            msg = await interaction.channel.send(f"❌ **{usuario.display_name}** saiu da lista.")
            await asyncio.sleep(5); await msg.delete()
        except Exception as e:
            await interaction.followup.send(f"❌ Erro: {e}", ephemeral=True)

@bot.event
async def on_ready():
    print(f'\n🚀 SISTEMA ONLINE! Bot {bot.user} pronto.')
    print("Mantenha o Play rodando. Vá ao Discord e digite: !mass")

@bot.command()
async def mass(ctx):
    global painel_ativo, mensagem_painel
    def check(m): return m.author == ctx.author and m.channel == ctx.channel
    try:
        await ctx.send(f"✍️ {ctx.author.mention}, digite o **canal de destino** (ex: #geral):")
        msg_c = await bot.wait_for('message', check=check, timeout=60.0)
        id_dest = "".join(re.findall(r'\d+', msg_c.content))
        canal_destino = bot.get_channel(int(id_dest)) if id_dest.isdigit() else None
        if not canal_destino:
            await ctx.send("❌ Canal inválido."); return
        await ctx.send("✍️ Digite a **Data/Hora Local de Saída**:")
        msg_d = await bot.wait_for('message', check=check, timeout=60.0)
        await ctx.send("✍️ Agora digite o **Nome do Conteúdo**:")
        msg_n = await bot.wait_for('message', check=check, timeout=60.0)
    except asyncio.TimeoutError:
        await ctx.send("⏱️ Tempo esgotado!"); return

    painel_ativo = PainelConvocacao(msg_n.content, msg_d.content)
    mensagem_painel = await canal_destino.send(embed=painel_ativo.gerar_embed_atualizado(), view=painel_ativo)
    await ctx.send("✅ Painel enviado!")

@bot.command(aliases=['call out'])
async def callout(ctx):
    global painel_ativo, mensagem_painel
    if not painel_ativo or not painel_ativo.confirmados:
        await ctx.send("⚠️ Nenhum painel ativo ou lista vazia."); return
    if not any(u.id == ctx.author.id for u in painel_ativo.confirmados): return

    pings = " ".join([f"<@{u.id}>" for u in painel_ativo.confirmados])
    await ctx.send(f"⚔️ **ALERTA DE CHAMADA!** ⚔️\n📦 **Conteúdo:** `{painel_ativo.nome_conteudo}`\n➔ {pings}")

    # Tranca o painel e desativa os botões em vez de apagar
    try:
        if mensagem_painel:
            painel_ativo.encerrado = True
            # Força a remoção física dos botões visíveis da mensagem
            await mensagem_painel.edit(embed=painel_ativo.gerar_embed_atualizado(), view=None)
            painel_ativo = None
            mensagem_painel = None
    except Exception as e:
        print(f"Erro ao trancar painel: {e}")

    try: await ctx.message.delete()
    except: pass

async def rodar():
    asyncio.ensure_future(bot.start(TOKEN))
    while True: await asyncio.sleep(1)

try: asyncio.get_event_loop().run_until_complete(rodar())
except KeyboardInterrupt: print("\n🛑 Desconectado.")
````
